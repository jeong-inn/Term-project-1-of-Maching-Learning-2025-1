```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import TimeSeriesSplit, GridSearchCV
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import StandardScaler
from xgboost import XGBRegressor

# 1. 헬퍼 함수 정의
def read_data(file_path):
    df = pd.read_csv(file_path)
    df['Date'] = pd.to_datetime(df['Date'], format='%Y-%m-%d')
    df.set_index('Date', inplace=True)
    return df

def add_features(df):
    # 타겟 변수 생성
    df['Target'] = df['Close'].shift(-1)
    
    # 추가 피처 생성
    df['MA_5'] = df['Close'].rolling(window=5).mean()
    df['Range'] = df['High'] - df['Low']
    df['High_Close_Ratio'] = (df['High'] - df['Close']) / df['Close']
    df['Close_Low_Ratio'] = (df['Close'] - df['Low']) / df['Close']
    df['Volume_Change'] = df['Volume'] / df['Volume'].shift(1)
    
    # 결측치 처리
    df['MA_5'] = df['MA_5'].ffill()
    df['Volume_Change'] = df['Volume_Change'].ffill()
    
    return df

# 2. 데이터 로드 및 전처리
df = read_data('/Users/jeonginn/Desktop/ML/kospi_train.csv')
df = add_features(df)
df = df.dropna()

# 피처 설정 (중요도가 낮은 피처 제외)
features = ['Open', 'Low', 'High', 'Close', 'Volume', 'MA_5']
X = df[features]
y = df['Target']

# 3. 피처 스케일링
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)  # NumPy 배열로 유지
X_scaled_df = pd.DataFrame(X_scaled, columns=features, index=X.index)  # 디버깅용 DataFrame

# 4. 모델 정의
# Random Forest 하이퍼파라미터 튜닝
rf_param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [5, 10],
    'min_samples_split': [5, 10]
}
rf = RandomForestRegressor(random_state=42)
rf_grid = GridSearchCV(rf, rf_param_grid, cv=TimeSeriesSplit(n_splits=3), scoring='neg_mean_squared_error')
rf_grid.fit(X_scaled, y)
best_rf = rf_grid.best_estimator_
print(f"Best Random Forest Params: {rf_grid.best_params_}")

models = {
    "LinearRegression": LinearRegression(),
    "RandomForest": best_rf,
    "XGBoost": XGBRegressor(n_estimators=100, max_depth=5, learning_rate=0.1, random_state=42)
}

# 5. 교차 검증
tscv = TimeSeriesSplit(n_splits=5)
results = {}

for name, model in models.items():
    rmses = []
    for train_idx, val_idx in tscv.split(X_scaled):
        X_train, X_val = X_scaled[train_idx], X_scaled[val_idx]
        y_train, y_val = y[train_idx], y[val_idx]
        
        model.fit(X_train, y_train)
        preds = model.predict(X_val)
        rmse = mean_squared_error(y_val, preds, squared=False)
        rmses.append(rmse)
    
    results[name] = {
        "mean_rmse": np.mean(rmses),
        "std_rmse": np.std(rmses)
    }

# 결과 출력
print("교차 검증 결과:")
for model_name, res in results.items():
    print(f"{model_name}: 평균 RMSE = {res['mean_rmse']:.2f}, 표준편차 = {res['std_rmse']:.2f}")

# 6. 피처 중요도 분석
# Linear Regression 회귀 계수
lr_model = models["LinearRegression"]
lr_model.fit(X_scaled, y)
lr_coefficients = pd.Series(lr_model.coef_, index=features)
print("\nLinear Regression 회귀 계수:")
print(lr_coefficients.sort_values(ascending=False))

# Random Forest 피처 중요도
rf_model = models["RandomForest"]
rf_importances = pd.Series(rf_model.feature_importances_, index=features)
print("\nRandom Forest 피처 중요도:")
print(rf_importances.sort_values(ascending=False))

# XGBoost 피처 중요도
xgb_model = models["XGBoost"]
xgb_model.fit(X_scaled, y)
xgb_importances = pd.Series(xgb_model.feature_importances_, index=features)
print("\nXGBoost 피처 중요도:")
print(xgb_importances.sort_values(ascending=False))

# 7. 테스트 데이터 준비 및 예측
df_test = read_data('/Users/jeonginn/Desktop/ML/kospi_test.csv')
df_test = add_features(df_test)
df_test = df_test.dropna()

# 테스트 데이터 준비
X_test = df_test[features]
y_test = df_test['Target']
X_test_scaled = scaler.transform(X_test)  # NumPy 배열로 변환

# 최적 모델로 예측 (Linear Regression 선택)
best_model = models["LinearRegression"]
best_model.fit(X_scaled, y)
predictions = best_model.predict(X_test_scaled)

# 테스트 RMSE 계산
test_rmse = mean_squared_error(y_test, predictions, squared=False)
print(f"\n테스트 데이터 RMSE: {test_rmse:.2f}")

# 예측 결과 출력
df_test['Predicted_Close'] = predictions
print("\n테스트 데이터 예측 결과 (상위 5개):")
print(df_test[['Close', 'Predicted_Close']].head())
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import StandardScaler
import yfinance as yf # yfinance 라이브러리 필요

# 기존 헬퍼 함수 사용 (앞서 정의한 read_data 함수)
def read_data(file_path):
    df = pd.read_csv(file_path)
    df['Date'] = pd.to_datetime(df['Date'], format='%Y-%m-%d')
    df.set_index('Date', inplace=True)
    return df

# --- 1. 코스피 학습 데이터 로드 ---
# 가정: kospi_train.csv 파일이 존재합니다.
df_train = read_data('/Users/jeonginn/Desktop/ML/kospi_train.csv')
print("Original KOSPI Train Data Head:")
print(df_train.head())
print("\nOriginal KOSPI Train Data Info:")
df_train.info()

# --- 2. 외부 데이터 가져오기 (학습 기간) ---
# 앞서 yfinance 코드를 실행했다고 가정하지만, 여기에서 다시 가져오는 코드를 포함합니다.
# 실제 과제에서는 이 부분을 앞선 yfinance 코드로 대체하면 됩니다.
start_date_train = df_train.index.min().strftime('%Y-%m-%d')
end_date_train = df_train.index.max().strftime('%Y-%m-%d')

# yfinance는 end date의 하루 전까지 데이터를 가져오므로, end date를 하루 뒤로 설정
end_date_train_yf = (df_train.index.max() + pd.Timedelta(days=1)).strftime('%Y-%m-%d')

sp500_data_train = yf.download('^GSPC', start=start_date_train, end=end_date_train_yf, progress=False)
krw_usd_data_train = yf.download('KRW=X', start=start_date_train, end=end_date_train_yf, progress=False)

print("\nFetched S&P 500 Train Data Head:")
print(sp500_data_train.head())
print("\nFetched KRW/USD Train Data Head:")
print(krw_usd_data_train.head())

# --- 3. 외부 데이터 전처리 및 병합 ---

# 사용할 컬럼 선택 및 이름 변경
# MultiIndex 컬럼에 접근할 때는 튜플을 사용해야 합니다.
# S&P 500은 'Close' 컬럼 사용 (Adj Close 없음)
# .rename()은 Series에도 적용 가능하며, .to_frame()으로 다시 DataFrame으로 변환
sp500_train_processed = sp500_data_train[('Close', '^GSPC')].rename('SP500_Close').to_frame()
krw_usd_train_processed = krw_usd_data_train[('Close', 'KRW=X')].rename('KRW_USD_Close').to_frame() # KRW=X는 'Close' 컬럼 사용

print("\nProcessed S&P 500 Train Data Head:")
print(sp500_train_processed.head())
print("\nProcessed KRW/USD Train Data Head:")
print(krw_usd_train_processed.head())


# 코스피 데이터에 외부 데이터 병합 (날짜 인덱스 기준)
# left merge를 사용하여 코스피 날짜는 모두 유지하고, 외부 데이터가 있는 날짜에만 추가
df_train_merged = df_train.copy()
df_train_merged = df_train_merged.merge(sp500_train_processed, left_index=True, right_index=True, how='left')
df_train_merged = df_train_merged.merge(krw_usd_train_processed, left_index=True, right_index=True, how='left')

print("\nMerged Train Data Head (with external features):")
print(df_train_merged.head())
print("\nMerged Train Data Info (check for NaNs after merge):")
df_train_merged.info()

# --- 4. 병합 후 결측치 처리 ---
# 외부 데이터의 휴일 등으로 인해 발생한 결측치는 이전 값으로 채우기 (ffill)
# 이제 컬럼 이름은 'SP500_Close', 'KRW_USD_Close' 입니다.
df_train_merged[['SP500_Close', 'KRW_USD_Close']] = df_train_merged[['SP500_Close', 'KRW_USD_Close']].ffill()

# 여전히 데이터 초반에 ffill로 채워지지 않는 결측치가 있을 수 있습니다.
# 이런 경우는 보통 해당 행을 제거하거나, 데이터 기간을 넉넉하게 가져와서 해결합니다.
# 여기서는 간단하게 남아있는 결측치 행은 제거하겠습니다.
# 제거 전에 얼마나 있는지 확인해볼 수도 있습니다: print(df_train_merged.isnull().sum())
df_train_merged.dropna(subset=['SP500_Close', 'KRW_USD_Close'], inplace=True)

print("\nTrain Data Head (after ffill and dropna):")
print(df_train_merged.head())
print("\nTrain Data Info (after ffill and dropna):")
df_train_merged.info()

# --- 5. 타겟 변수 및 파생 피처 생성 (병합된 데이터 기준) ---
# 타겟 변수 생성 (다음 날 종가)
df_train_merged['Target'] = df_train_merged['Close'].shift(-1)

# 앞서 정의한 파생 피처 추가 (이제 병합된 데이터프레임을 기준으로 계산)
# 여기서는 MA_5만 포함하겠지만, 앞서 계산했던 Range, High_Close_Ratio 등도 추가 가능
df_train_merged['MA_5'] = df_train_merged['Close'].rolling(window=5).mean().ffill()
# 필요한 경우 다른 파생 피처들을 여기에 추가하세요.
# df_train_merged['Range'] = df_train_merged['High'] - df_train_merged['Low']
# df_train_merged['High_Close_Ratio'] = (df_train_merged['High'] - df_train_merged['Close']) / df_train_merged['Close']
# df_train_merged['Volume_Change'] = df_train_merged['Volume'] / df_train_merged['Volume'].shift(1).ffill()


# --- 6. 최종 데이터셋 준비 (결측치 제거) ---
# 타겟 변수 생성으로 인해 마지막 행에 NaN이 생기므로 제거합니다.
# 파생 피처 생성 시 초반 N개 행에 NaN이 생길 수 있으므로 함께 제거합니다.
df_train_final = df_train_merged.dropna().copy()

print("\nFinal Train Data Head (with Target and MA_5, NaNs dropped):")
print(df_train_final.head())
print("\nFinal Train Data Info:")
df_train_final.info()

# --- 7. 피처(X)와 타겟(y) 분리 ---
# 사용할 피처 목록 정의 (기존 + 파생 + 외부)
features_extended = ['Open', 'Low', 'High', 'Close', 'Volume', 'MA_5', 'SP500_Close', 'KRW_USD_Close']
# 만약 다른 파생 피처를 추가했다면 features_extended 리스트에 포함시키세요.

X_train = df_train_final[features_extended]
y_train = df_train_final['Target']

print(f"\nShape of X_train: {X_train.shape}")
print(f"Shape of y_train: {y_train.shape}")

# --- 8. 피처 스케일링 (학습 데이터에 fit) ---
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)


# --- 9. 테스트 데이터 준비 (학습 데이터와 동일한 과정 적용) ---

# 코스피 테스트 데이터 로드
df_test = read_data('/Users/jeonginn/Desktop/ML/kospi_test.csv')

# 외부 데이터 가져오기 (테스트 기간)
start_date_test = df_test.index.min().strftime('%Y-%m-%d')
end_date_test = df_test.index.max().strftime('%Y-%m-%d')
end_date_test_yf = (df_test.index.max() + pd.Timedelta(days=1)).strftime('%Y-%m-%d')

sp500_data_test = yf.download('^GSPC', start=start_date_test, end=end_date_test_yf, progress=False)
krw_usd_data_test = yf.download('KRW=X', start=start_date_test, end=end_date_test_yf, progress=False)


# 외부 데이터 전처리 및 병합 (테스트 데이터)
# 테스트 데이터도 동일하게 MultiIndex 접근 사용
sp500_test_processed = sp500_data_test[('Close', '^GSPC')].rename('SP500_Close').to_frame()
krw_usd_test_processed = krw_usd_data_test[('Close', 'KRW=X')].rename('KRW_USD_Close').to_frame()

df_test_merged = df_test.copy()
df_test_merged = df_test_merged.merge(sp500_test_processed, left_index=True, right_index=True, how='left')
df_test_merged = df_test_merged.merge(krw_usd_test_processed, left_index=True, right_index=True, how='left')

# 병합 후 결측치 처리 (테스트 데이터 - 학습 데이터와 동일한 방식)
df_test_merged[['SP500_Close', 'KRW_USD_Close']] = df_test_merged[['SP500_Close', 'KRW_USD_Close']].ffill()
df_test_merged.dropna(subset=['SP500_Close', 'KRW_USD_Close'], inplace=True) # 초반 결측치 제거

# 타겟 변수 및 파생 피처 생성 (테스트 데이터)
df_test_merged['Target'] = df_test_merged['Close'].shift(-1)
df_test_merged['MA_5'] = df_test_merged['Close'].rolling(window=5).mean().ffill()
# 여기에 학습 데이터에 추가한 다른 파생 피처들도 동일하게 추가해야 합니다.

# 최종 테스트 데이터셋 준비 (결측치 제거)
df_test_final = df_test_merged.dropna().copy()

# 피처(X)와 타겟(y) 분리 (테스트 데이터) - 학습 데이터와 동일한 피처 목록 사용
X_test = df_test_final[features_extended]
y_test = df_test_final['Target']

print(f"\nShape of X_test: {X_test.shape}")
print(f"Shape of y_test: {y_test.shape}")

# 피처 스케일링 (테스트 데이터 - 학습 데이터에서 fit한 scaler 사용)
X_test_scaled = scaler.transform(X_test)

# --- 10. 모델 학습 및 평가 (확장된 피처셋 사용) ---
# 예시로 Linear Regression 사용. 앞서 교차 검증으로 찾은 최적 모델 사용 가능.
model = LinearRegression()

# 학습 데이터 전체로 모델 재학습
model.fit(X_train_scaled, y_train)

# 테스트 데이터로 예측
predictions = model.predict(X_test_scaled)

# 테스트 RMSE 계산
test_rmse = mean_squared_error(y_test, predictions, squared=False)
print(f"\n모델 (Linear Regression) 테스트 데이터 RMSE (확장된 피처): {test_rmse:.2f}")

# 예측 결과 출력
#df_test_final['Predicted_Close'] = predictions
df_test_final.loc[:, 'Predicted_Close'] = predictions

print("\n테스트 데이터 예측 결과 (확장된 피처, 상위 5개):")
print(df_test_final[['Close', 'Predicted_Close']].head())
```
